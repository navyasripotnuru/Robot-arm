```python

import math
import matplotlib.pyplot as plt

print("======================================")
print("     ROBOT ARM POSITION CALCULATOR")
print("======================================")

# Input values
L1 = float(input("Enter length of first arm (L1): "))
L2 = float(input("Enter length of second arm (L2): "))
theta1 = float(input("Enter angle of first joint (theta1): "))
theta2 = float(input("Enter angle of second joint (theta2): "))

# Convert degrees to radians
t1 = math.radians(theta1)
t2 = math.radians(theta2)

# Calculate first joint position
x1 = L1 * math.cos(t1)
y1 = L1 * math.sin(t1)

# Calculate end-effector position
x2 = x1 + L2 * math.cos(t1 + t2)
y2 = y1 + L2 * math.sin(t1 + t2)

# Display calculations
print("\n========== RESULT ==========")
print("First Joint Position:")
print("X1 =", round(x1, 2))
print("Y1 =", round(y1, 2))

print("\nRobot Arm End Position:")
print("X2 =", round(x2, 2))
print("Y2 =", round(y2, 2))

# Plot robot arm
plt.figure(figsize=(7, 7))

# Robot arm
plt.plot([0, x1, x2], [0, y1, y2], marker='o', linewidth=4)

# Mark points
plt.text(0, 0, "  Base")
plt.text(x1, y1, "  Joint")
plt.text(x2, y2, "  End Point")

# X and Y axes
plt.axhline(0, linewidth=1)
plt.axvline(0, linewidth=1)

# Graph settings
plt.xlabel("X Position")
plt.ylabel("Y Position")
plt.title("Robot Arm Position Calculator")
plt.grid(True)
plt.axis("equal")

plt.show()

print("\nRobot arm position calculated successfully!")
```

```python

import math
import matplotlib.pyplot as plt

print("ROBOT ARM POSITION CALCULATOR")
print("------------------------------")

while True:

    # Input
    L1 = float(input("\nEnter length of first arm: "))
    L2 = float(input("Enter length of second arm: "))
    theta1 = float(input("Enter first joint angle (degrees): "))
    theta2 = float(input("Enter second joint angle (degrees): "))

    # Convert degrees to radians
    angle1 = math.radians(theta1)
    angle2 = math.radians(theta2)

    # Position of first joint
    x1 = L1 * math.cos(angle1)
    y1 = L1 * math.sin(angle1)

    # Position of end point
    x2 = x1 + L2 * math.cos(angle1 + angle2)
    y2 = y1 + L2 * math.sin(angle1 + angle2)

    # Display result
    print("\nRESULT")
    print("------")
    print("First Joint Position:")
    print("X1 =", round(x1, 2))
    print("Y1 =", round(y1, 2))

    print("\nRobot Arm End Position:")
    print("X =", round(x2, 2))
    print("Y =", round(y2, 2))

    # Draw robot arm
    plt.figure(figsize=(6, 6))
    plt.plot([0, x1, x2], [0, y1, y2], marker='o', linewidth=4)

    plt.scatter(0, 0, s=100, label="Base")
    plt.scatter(x1, y1, s=100, label="Joint")
    plt.scatter(x2, y2, s=100, label="End Point")

    plt.xlabel("X Position")
    plt.ylabel("Y Position")
    plt.title("Robot Arm Position")
    plt.grid(True)
    plt.axis("equal")
    plt.legend()
    plt.show()

    # Continue or stop
    choice = input("\nDo you want to calculate again? (yes/no): ")

    if choice.lower() != "yes":
        print("\nThank you!")
        print("Robot Arm Position Calculator completed.")
        break
```
